```python
# necessary components to run tube_calc.ipynb
import numpy as np
from numpy import pi, sin, cos
import math
```

```python
# parameters 

P = 5000       # newtons, need further clarification on thsi value

r_p = 672.08          #!!!! d to p length mm
r_hand = 750.633     #!!!!!!!! d to hand length in mm
degp = 21.4              # !! angle between p force directino and DP
degh = 33.5        # !! angle between hand force direction and DH

r_b = 30           # a to b, in mm

alpha = 62.6     # degress
beta = 91.1         # degress

gamma = 75     # seat tube angle, deg
phi = 66         # head tube angle, deg

# ------------------------------------- #

# unit conversion
#need all the unit conversions here, mm to meters, deg to radians

k = 1 / 1000             # constants
c = pi / 180

r_p = r_p * k         # to meters
r_hand = r_hand * k

r_b = r_b * k

alpha = alpha * c     # to radians
beta = beta * c

gamma = gamma * c
phi = phi * c

print(r_p,r_hand,r_b)
print(alpha,beta,gamma,phi)
```

```python
#1 global
   # point a = origin

torque_p = P * r_p *sin(degp * c) 
# torque_p = torque_f = F_hand * r_hand * sin(degh * c)

F_hand = torque_p / (r_hand * sin(degh * c))
D_x = P - F_hand # pivot pushes on the traigne, to right


#2 fork
   # point a = origin

# F_hand + F_b + F_a + P  = 0, assuming left is positive
# F_hand * r_hand + F_b * r_b + P * r_p = 0

F_b = (F_hand * r_hand + P * r_p) / r_b  # to left
F_a = F_b - F_hand - P      # algining with fb direction

#3 frame

F_net = F_a + F_b                # force on the triangle from the fork section, to the left

# F_tt * sin (beta) + F_dt * sin(alpha) = F_net
# F_tt * cos (beta) = F_dt * cos(alpha) , mangituede

F_tt = F_net - F_dt * sin(alpha) / sin( beta) # to left
F_dt = -F_tt * cos(beta) / cos(alpha) # to left

#4 seat tube

# summing x direction

F_st = ( F_dt * sin(alpha) + F_tt * sin(beta) - D_x) / sin(gamma-phi)

print('general:' , P, F_hand, D_x)
print('fork: ', F_a, F_b)
print('frame: ', F_tt, F_dt, F_st)
```

```python
# goal tube diameter

# need params

wall = 0.75 #mm

fail = 290000000 #pa

FOS = 2

allow = fail / FOS



# result:

wall_m = wall * k   # m

F = np.array([F_tt, F_dt, F_st])

area = np.abs(F) / allow
area_mm2 = area * 1000000

id_raw = area_mm2 / (pi * wall) - wall
id = np.maximum(id_raw, 0)
od = id + 2 * wall
valid_hollow = id_raw >= 0

# print("F_tt, F_dt, F_st =", F)
# print("area (m^2) =", area)
print("area (mm^2) =", area_mm2)
# print("id raw (mm) =", id_raw)
print("valid hollow tube =", valid_hollow)
print("id (mm) =", id)
print("od (mm) =", od)
```
