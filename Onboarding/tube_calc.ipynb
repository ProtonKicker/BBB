```python
# necessary components to run tube_calc.ipynb
import numpy as np
from numpy import pi, sin, cos
```

```python
# parameters 

import math


P = 5000       # newtons, need further clarification on thsi value

r_p = 672.08          # a to p length mm
r_hand = 750.633 

r_b = 30           # a to b, in mm

alpha = 62.6     # degress
beta = 91.1         # degress

gamma = 75     # seat tube angle, deg
phi = 66         # head tube angle, deg

# ------------------------------------- #

# unit conversion
#need all the unit conversions here, mm to meters, deg to radians

k = 1 / 1000             # constants
c = pi / 180

r_p = r_p * k         # to meters
r_hand = r_hand * k

r_b = r_b * k

alpha = alpha * c     # to radians
beta = beta * c

gamma = gamma * c
phi = phi * c

print(r_p,r_hand,r_b)
print(alpha,beta,gamma,phi)
```

```python
#1 global
   # point a = origin

tempP = P * sin(21.4 * c)  
tempR = r_hand * sin(33.5 *c)

F_hand = r_p * tempP / tempR
D_x = F_hand - P


#2 fork
   # point a = origin

# F_hand + F_b + F_a + P  = 0, assuming left is positive
# F_hand * r_hand + F_b * r_b + P * r_p = 0

F_b = (F_hand * r_hand + P * r_p) / r_b  # to left
F_a = F_hand - F_b - P      # to right

#3 frame

F_net = ( - 1 ) * ( F_a + F_b )                # force on the triangle from the fork section, to the left

# F_tt * sin (beta) + F_dt * sin(alpha) + F_net = 0
# F_tt * cos (beta) + F_dt * cos(alpha) = 0

F_tt = F_net * cos(alpha) / ( sin(alpha - beta))
F_dt = -F_tt * cos(beta) / cos(alpha)

#4 seat tube

# D_x + F_dt * sin(alpha) + F-tt * sin(beta) - F_st * sin(gamma - phi)

F_st = ( D_x + F_dt * sin(alpha) + F_tt * sin(beta) ) / sin(beta + phi - gamma)

print(tempP)
print(F_hand, D_x)
print(F_a, F_b, F_tt, F_dt, F_st)
```

```python
# goal tube diameter

# need params

wall = 0.75 #mm

fail = 290000000 #pa

FOS = 2

allow = fail / FOS



# result:

wall = wall * k   # m

F = np.array([F_tt, F_dt, F_st])

area = np.abs(F) / allow
   
od = area / (pi * wall) + wall # taking mm^2 for area,  area = pi * (od^2 - (od - wall)^2)

od = od * 1000    # od in milimeters

print("F_tt, F_dt, F_st =", F)
print("area =", area)
print("od (mm) =", od)
```
